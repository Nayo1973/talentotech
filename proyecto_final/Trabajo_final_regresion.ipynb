## Modelo de Regresión Múltiple para Predecir la Depresión

Vamos a construir un modelo de regresión lineal múltiple donde la variable objetivo será 'Depresion' y el resto de las variables serán las características. Primero, necesitamos preparar los datos, manejar las variables categóricas, dividir el dataset y luego entrenar y evaluar el modelo.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# 1. Preparación de datos
# Definir la variable objetivo (y) como 'Ansiedad' y las características (X)
# Excluir 'ID_Persona' y la nueva variable objetivo 'Ansiedad' de las características
X = df.drop(columns=['ID_Persona', 'Ansiedad'])
y = df['Ansiedad']

# Identificar columnas categóricas para One-Hot Encoding
categorical_features = X.select_dtypes(include=['object']).columns
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True)

display(X_encoded.head())
print(f"Dimensiones de X_encoded: {X_encoded.shape}")
```

### División de Datos en Conjuntos de Entrenamiento y Prueba

Dividimos los datos para poder entrenar el modelo con un subconjunto y evaluar su rendimiento con datos no vistos.

```python
# 2. División de datos
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print(f"Tamaño del conjunto de entrenamiento X: {X_train.shape}")
print(f"Tamaño del conjunto de prueba X: {X_test.shape}")
print(f"Tamaño del conjunto de entrenamiento y: {y_train.shape}")
print(f"Tamaño del conjunto de prueba y: {y_test.shape}")
```

### Creación y Entrenamiento del Modelo de Regresión Lineal

Ahora instanciamos y entrenamos el modelo de Regresión Lineal con los datos de entrenamiento.

```python
# 3. Creación y entrenamiento del modelo
model = LinearRegression()
model.fit(X_train, y_train)

print("Modelo de Regresión Lineal entrenado con éxito.")
```

### Evaluación del Modelo

Finalmente, evaluamos el rendimiento del modelo utilizando métricas comunes como R-cuadrado, Error Absoluto Medio (MAE) y Error Cuadrático Medio (MSE) en el conjunto de prueba.

```python
# 4. Evaluación del modelo
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"Error Absoluto Medio (MAE): {mae:.4f}")
print(f"Error Cuadrático Medio (MSE): {mse:.4f}")
print(f"Raíz del Error Cuadrático Medio (RMSE): {rmse:.4f}")
print(f"Coeficiente de Determinación (R^2): {r2:.4f}")
```

### Análisis de Coeficientes del Modelo

Vamos a examinar los coeficientes de las características para entender su impacto en la variable objetivo 'Depresion'. Un coeficiente más grande (en valor absoluto) indica un mayor impacto.

```python
# Obtener los coeficientes y los nombres de las características
coefficients = model.coef_
feature_names = X_encoded.columns

# Crear un DataFrame para una mejor visualización
coef_df = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})

# Ordenar por el valor absoluto del coeficiente para ver los de mayor impacto
coef_df['Absolute_Coefficient'] = abs(coef_df['Coefficient'])
coef_df = coef_df.sort_values(by='Absolute_Coefficient', ascending=False)

display(coef_df)
```

## Reestructuración del Dataset para Reentrenamiento

Ahora, vamos a seleccionar las variables específicas que solicitaste para reentrenar el modelo. Estas variables son: `Actividad_Fisica_Low`, `Ocupacion_Unemployed`, `Ansiedad`, `Nivel_Estres`, `Fumador`, `Alcohol`, `Agotamiento`, `Horas_Sueño`, y `Puntuacion_Interaccion_Social`.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd # Ensure pandas is imported

# 1. Preparación de datos - Re-defining X, y, X_encoded for the new target 'Ansiedad'
# Definir la variable objetivo (y) como 'Ansiedad' y las características (X)
# Excluir 'ID_Persona' y la nueva variable objetivo 'Ansiedad' de las características
X = df.drop(columns=['ID_Persona', 'Ansiedad'])
y = df['Ansiedad']

# Identificar columnas categóricas para One-Hot Encoding
categorical_features = X.select_dtypes(include=['object']).columns
X_encoded = pd.get_dummies(X, columns=categorical_features, drop_first=True)

selected_features = [
    'Actividad_Fisica_Low',
    'Ocupacion_Unemployed',
    'Nivel_Estres',
    'Agotamiento',
    'Horas_Sueño',
    'Puntuacion_Interaccion_Social',
    'Depresion' # 'Depresion' ahora puede ser una característica
]

# Asegurarse de que todas las características seleccionadas existan en X_encoded
missing_features = [f for f in selected_features if f not in X_encoded.columns]
if missing_features:
    print(f"Advertencia: Las siguientes características no se encontraron en el dataset codificado: {missing_features}")

X_selected = X_encoded[selected_features]

print(f"Nuevas dimensiones de X_selected: {X_selected.shape}")
display(X_selected.head())
```

### Nueva División de Datos y Reentrenamiento del Modelo

Con el dataset reestructurado, vamos a dividir los datos nuevamente y entrenar un nuevo modelo de regresión lineal.

```python
# 2. División de datos con las características seleccionadas
X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(X_selected, y, test_size=0.2, random_state=42)

print(f"Tamaño del nuevo conjunto de entrenamiento X: {X_train_new.shape}")
print(f"Tamaño del nuevo conjunto de prueba X: {X_test_new.shape}")

# 3. Creación y entrenamiento del nuevo modelo
model_new = LinearRegression()
model_new.fit(X_train_new, y_train_new)

print("Nuevo Modelo de Regresión Lineal entrenado con éxito usando las características seleccionadas.")
```

### Evaluación del Nuevo Modelo

Ahora evaluaremos el rendimiento del modelo entrenado con el subconjunto de características.

```python
# 4. Evaluación del nuevo modelo
y_pred_new = model_new.predict(X_test_new)

mae_new = mean_absolute_error(y_test_new, y_pred_new)
mse_new = mean_squared_error(y_test_new, y_pred_new)
rmse_new = np.sqrt(mse_new)
r2_new = r2_score(y_test_new, y_pred_new)

print(f"Error Absoluto Medio (MAE) del nuevo modelo: {mae_new:.4f}")
print(f"Error Cuadrático Medio (MSE) del nuevo modelo: {mse_new:.4f}")
print(f"Raíz del Error Cuadrático Medio (RMSE) del nuevo modelo: {rmse_new:.4f}")
print(f"Coeficiente de Determinación (R^2) del nuevo modelo: {r2_new:.4f}")
```

### Análisis de Coeficientes del Nuevo Modelo

Finalmente, examinemos los coeficientes del modelo entrenado con las características seleccionadas para entender su impacto en la 'Depresion'.

```python
# Obtener los coeficientes y los nombres de las nuevas características
coefficients_new = model_new.coef_
feature_names_new = X_selected.columns

# Crear un DataFrame para una mejor visualización
coef_df_new = pd.DataFrame({'Feature': feature_names_new, 'Coefficient': coefficients_new})

# Ordenar por el valor absoluto del coeficiente para ver los de mayor impacto
coef_df_new['Absolute_Coefficient'] = abs(coef_df_new['Coefficient'])
coef_df_new = coef_df_new.sort_values(by='Absolute_Coefficient', ascending=False)

display(coef_df_new)
```

#### Interpretación de los Coeficientes:

*   **Magnitud:** Un valor absoluto más alto en el coeficiente indica que esa característica tiene un mayor impacto en la variable 'Depresion'.
*   **Signo (positivo/negativo):**
    *   Un coeficiente **positivo** significa que a medida que esa característica aumenta, la 'Depresion' tiende a aumentar (manteniendo las otras variables constantes).
    *   Un coeficiente **negativo** significa que a medida que esa característica aumenta, la 'Depresion' tiende a disminuir (manteniendo las otras variables constantes).

En un modelo de regresión lineal simple como este, el coeficiente representa el cambio promedio en la variable dependiente (Depresion) por cada unidad de cambio en la variable independiente (característica), asumiendo que todas las demás variables se mantienen constantes.
